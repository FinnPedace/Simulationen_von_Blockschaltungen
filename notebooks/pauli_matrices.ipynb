# Pauli-Matrizen

Dieses Notebook testet die drei Pauli-Matrizen und überprüft numerisch die Relation `sigma_x @ sigma_y = i * sigma_z`.

```python
import sys
from pathlib import Path

import numpy as np

for source_directory in (Path.cwd() / 'src', Path.cwd().parent / 'src'):
    if (source_directory / 'Pauli_matrices.py').exists():
        sys.path.insert(0, str(source_directory))
        break

from Pauli_matrices import pauli_x, pauli_y, pauli_z
```

```python
sigma_x = pauli_x()
sigma_y = pauli_y()
sigma_z = pauli_z()
print('sigma_x:')
print(sigma_x)
print('sigma_y:')
print(sigma_y)
print('sigma_z:')
print(sigma_z)
```

```python
left_side = sigma_x @ sigma_y
right_side = 1j * sigma_z

print('sigma_x @ sigma_y:')
print(left_side)
print('i * sigma_z:')
print(right_side)
print('Relation holds:', np.allclose(left_side, right_side))

assert np.allclose(left_side, right_side)
```
